# identityexcess: minimal reproducible example
Companion notebook for *Identity Programme vs Intra-program Remodelling in load-bearing fibrotic tissues*

This notebook demonstrates:
1. Rank-first UCell signature scoring
2. Identity-excess calculation
3. Permutation TOST equivalence test


```python
import numpy as np
from identityexcess import compute_identity_excess, tost_equivalence_test, ucell_rank_score
import matplotlib.pyplot as plt
```

## 1. Simulate single-cell expression matrix
Simulate 200 cells × 1000 genes, signature gene indices: 0–99

```python
rng = np.random.default_rng(42)
mat = rng.normal(loc=0, scale=1, size=(200, 1000))
sig_genes = np.arange(0,100)
scores = ucell_rank_score(mat, sig_genes)
print('Signature score shape:', scores.shape)
```

## 2. Identity-excess & TOST test
Two score vectors: identity programme, remodelling programme

```python
score_id = np.array([0.21,0.33,0.18,0.42,0.30,0.27,0.35,0.29])
score_rem = np.array([0.11,0.15,0.09,0.20,0.14,0.12,0.17,0.13])

excess = compute_identity_excess(score_id, score_rem)
res = tost_equivalence_test(score_id, score_rem, eq_margin=0.05, n_perm=2000)

print(f'Identity excess = {excess:.3f}')
print('TOST result:', res)
```

## 3. Simple visualization

```python
plt.boxplot([score_id, score_rem], labels=['Identity programme','Remodelling programme'])
plt.ylabel('Signature score')
plt.title(f'Identity excess = {excess:.3f}')
plt.show()
```

## Notes for reviewers
- `ucell_rank_score` implements rank-first scoring (per-cell global ranking before signature averaging, matching stage4 pipeline)
- TOST uses permutation null distribution, not parametric assumption
- Real analysis uses frozen 100-gene MMP2hi_FIB signature from the manuscript
